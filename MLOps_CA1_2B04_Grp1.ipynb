````python
def parse_batch_dates(date_values):
    """Parse the source date formats and leave invalid dates missing."""
    return pd.to_datetime(date_values, format="mixed", errors="coerce")


# Count non-missing inputs that became missing during date parsing.
def count_unparseable_dates(date_values):
    return (date_values.notna() & parse_batch_dates(date_values).isna()).sum()


def summarise_invalid_values(customers, orders, labels):
    # Convert item counts to numbers so invalid text and fractional counts can be flagged.
    item_counts = pd.to_numeric(orders["num_items_ordered"], errors="coerce")
    invalid_item_counts = orders["num_items_ordered"].notna() & (
        item_counts.isna() | item_counts.le(0) | item_counts.mod(1).ne(0)
    )

    def count_invalid_amounts(raw_amounts):
        amounts = pd.to_numeric(raw_amounts, errors="coerce")
        return (raw_amounts.notna() & (
            amounts.isna() | amounts.lt(0) | amounts.eq(float("inf"))
        )).sum()

    return pd.Series({
        "Unparseable order dates": count_unparseable_dates(orders["order_date"]),
        "Unparseable first-order dates": count_unparseable_dates(customers["first_order_datetime"]),
        "Invalid or missing fraud labels": (
            ~pd.to_numeric(labels["is_fraud"], errors="coerce").isin([0, 1])
        ).sum(),
        "Invalid item counts": invalid_item_counts.sum(),
        "Invalid order values": count_invalid_amounts(orders["order_value"]),
        "Invalid refund values": count_invalid_amounts(orders["refund_value"]),
    }, name="Rows").to_frame()


summarise_invalid_values(customers_base, orders_base, labels_base)
````

## 3️⃣Preprocessing and Feature Engineering

### 3.1 Clean the source tables 

`clean_batch` applies the same rules to any batch. It returns three cleaned tables and an audit. Optional missing values stay missing. Rows with unusable keys, order dates, or labels are excluded and counted. If rows share a key and disagree, all conflicting rows are excluded.

````python
def clean_table_keys(raw_table, key_columns):
    table = raw_table.copy()
    table["country_code"] = normalise_country_codes(table["country_code"].astype("string"))
    for column in key_columns:
        if column != "country_code":
            table[column] = table[column].astype("string").str.strip()
    invalid_key = ~table["country_code"].str.fullmatch(r"[A-Z]{2}", na=False)
    invalid_key |= table[key_columns].fillna("").eq("").any(axis=1)
    return table.loc[~invalid_key].copy(), {
        "Input rows": len(raw_table),
        "Invalid key rows": int(invalid_key.sum()),
    }


def resolve_duplicate_keys(table, key_columns, audit):
    # Compare standardised values before deciding whether records disagree.
    distinct_rows = table.drop_duplicates()
    conflicting_keys = distinct_rows.duplicated(key_columns, keep=False)
    audit["Repeated rows removed"] = len(table) - len(distinct_rows)
    audit["Conflicting key rows"] = int(conflicting_keys.sum())
    return distinct_rows.loc[~conflicting_keys].copy(), audit


def clean_customer_records(raw_customers):
    customers, audit = clean_table_keys(raw_customers, ["country_code", "customer_id"])
    # Convert the supported verification strings to nullable booleans.
    verification = customers["mobile_verified"].astype("string").str.strip().str.lower()
    mapped = verification.map({"true": True, "yes": True, "false": False})
    audit["Unexpected verification values"] = int((verification.notna() & mapped.isna()).sum())
    customers["mobile_verified"] = mapped.astype("boolean")
    # Parse first-order dates and count non-missing values that failed to parse.
    first_order = parse_batch_dates(customers["first_order_datetime"])
    audit["Unparseable first-order dates"] = int(
        (customers["first_order_datetime"].notna() & first_order.isna()).sum()
    )
    customers["first_order_datetime"] = first_order
    return resolve_duplicate_keys(customers, ["country_code", "customer_id"], audit)


def clean_order_records(raw_orders):
    orders, audit = clean_table_keys(raw_orders, ["country_code", "order_id"])
    # Parse order dates, keep the calendar day, and filter rows without a usable date.
    orders["order_date"] = parse_batch_dates(orders["order_date"]).dt.normalize()
    invalid_date = orders["order_date"].isna()
    audit["Missing or invalid order dates"] = int(invalid_date.sum())
    orders = orders.loc[~invalid_date].copy()

    # Convert item counts to numbers so invalid text and fractional counts can be flagged.
    item_counts = pd.to_numeric(orders["num_items_ordered"], errors="coerce")
    valid_counts = item_counts.gt(0) & item_counts.mod(1).eq(0) & item_counts.lt(2**63)
    audit["Invalid item counts set to missing"] = int(
        (orders["num_items_ordered"].notna() & ~valid_counts).sum()
    )
    orders["num_items_ordered"] = item_counts.where(valid_counts).astype("Int64")
    # Keep finite, non-negative amounts and set invalid amounts to missing.
    for column in ("order_value", "refund_value"):
        amounts = pd.to_numeric(orders[column], errors="coerce")
        valid_amount = amounts.ge(0) & amounts.lt(float("inf"))
        audit[f"Invalid {column} set to missing"] = int(
            (orders[column].notna() & ~valid_amount).sum()
        )
        orders[column] = amounts.where(valid_amount)
    # Standardise category casing and whitespace, then treat blanks as missing.
    for column in ("payment_method", "collect_type"):
        orders[column] = orders[column].astype("string").str.strip().str.lower().replace("", pd.NA)
    return resolve_duplicate_keys(orders, ["country_code", "order_id"], audit)


def clean_label_records(raw_labels):
    labels, audit = clean_table_keys(raw_labels, ["country_code", "order_id"])
    labels["customer_id"] = labels["customer_id"].astype("string").str.strip()
    missing_customer = labels["customer_id"].fillna("").eq("")
    # Convert the target to a number and keep only 0 or 1.
    fraud_values = pd.to_numeric(labels["is_fraud"], errors="coerce")
    invalid_label = ~fraud_values.isin([0, 1])
    audit["Missing customer IDs"] = int(missing_customer.sum())
    audit["Invalid or missing fraud labels"] = int(invalid_label.sum())
    valid_rows = ~(missing_customer | invalid_label)
    labels = labels.loc[valid_rows].copy()
    labels["is_fraud"] = fraud_values.loc[valid_rows].astype("int8")
    return resolve_duplicate_keys(labels, ["country_code", "order_id"], audit)


def clean_batch(raw_customers, raw_orders, raw_labels):
    """Apply the same cleaning rules to any batch loaded from CSV."""
    # Run the three cleaners and collect their row counts in one audit table.
    results = {
        "Customers": clean_customer_records(raw_customers),
        "Orders": clean_order_records(raw_orders),
        "Labels": clean_label_records(raw_labels),
    }
    cleaned = {name: table for name, (table, _) in results.items()}
    audit = pd.DataFrame.from_dict({
        name: {**counts, "Cleaned rows": len(table)}
        for name, (table, counts) in results.items()
    }, orient="index").astype("Int64")
    return cleaned, audit
````

### 3.2 Join labels to orders and customers 

First, join labels to orders on `(country_code, order_id)`. Then join to customers on `(country_code, customer_id)`. Each label can match one order, and several labels can match the same customer. Left joins keep every label visible in the audit.

````python
def join_labelled_orders(labels, orders, customers):
    # Keep every label and record whether each of its two joins found a match.
    joined = labels.merge(
        orders, on=["country_code", "order_id"], how="left",
        validate="one_to_one", indicator="order_match",
    ).merge(
        customers, on=["country_code", "customer_id"], how="left",
        validate="many_to_one", indicator="customer_match",
    )
    # Keep fully matched records for features and return the remaining label keys for review.
    has_order = joined["order_match"].eq("both")
    has_customer = joined["customer_match"].eq("both")
    matched = joined.loc[has_order & has_customer].drop(
        columns=["order_match", "customer_match"]
    )
    unmatched = joined.loc[~(has_order & has_customer), [
        "country_code", "order_id", "customer_id", "order_match", "customer_match"
    ]].copy()
    return matched, unmatched


matched_orders, unmatched_labels = join_labelled_orders(
    cleaned_tables["Labels"], cleaned_tables["Orders"], cleaned_tables["Customers"]
)
assert len(matched_orders) + len(unmatched_labels) == len(cleaned_tables["Labels"])
pd.Series({
    "Fully matched": len(matched_orders),
    "Missing order only": (
        unmatched_labels["order_match"].eq("left_only")
        & unmatched_labels["customer_match"].eq("both")
    ).sum(),
    "Missing customer only": (
        unmatched_labels["customer_match"].eq("left_only")
        & unmatched_labels["order_match"].eq("both")
    ).sum(),
    "Missing both": (
        unmatched_labels["order_match"].eq("left_only")
        & unmatched_labels["customer_match"].eq("left_only")
    ).sum(),
}, name="Label rows").to_frame()
````

**Conclusion.** 949,872 labels matched both tables. The other 446 remain in the join audit and are excluded from the model dataset.

### 3.3 Build features for order review 

| Feature | Why it may help |
|---|---|
| Days since first order | Measures how recently the customer started ordering. |
| Value per item | Measures how expensive the order is relative to its size. |
| Order weekday | Captures possible differences in ordering patterns across the week. |

Use order details available before shipment. Refunds and the customer snapshot fields stay out of the model inputs because they may include information from after the order. The customer ID and order ID are retained for tracing and split checks.

````python
# List the columns intended as model inputs, excluding IDs and the target.
MODEL_FEATURE_COLUMNS = [
    "country_code", "collect_type", "payment_method", "order_value",
    "num_items_ordered", "days_since_first_order", "value_per_item", "order_weekday",
]


def build_order_features(labelled_orders):
    model_data = labelled_orders[[
        "country_code", "order_id", "customer_id", "order_date",
        "collect_type", "payment_method", "order_value", "num_items_ordered", "is_fraud",
    ]].copy()
    # Compare calendar days and leave negative or unknown elapsed times missing.
    first_order_day = labelled_orders["first_order_datetime"].dt.normalize()
    days_since_first_order = (model_data["order_date"] - first_order_day).dt.days
    model_data["days_since_first_order"] = days_since_first_order.where(
        days_since_first_order.ge(0)
    ).astype("Int64")
    # Use positive item counts as the denominator so missing or zero counts yield no ratio.
    positive_item_count = model_data["num_items_ordered"].where(
        model_data["num_items_ordered"].gt(0)
    )
    model_data["value_per_item"] = model_data["order_value"].div(positive_item_count)
    model_data["order_weekday"] = model_data["order_date"].dt.day_name()
    # Represent missing payment methods as a category in the model dataset.
    model_data["payment_method"] = model_data["payment_method"].fillna("Unknown")
    return model_data


model_data = build_order_features(matched_orders)
assert len(model_data) == len(matched_orders)
assert model_data["days_since_first_order"].dropna().ge(0).all()
assert model_data["value_per_item"].dropna().ge(0).all()
model_data[MODEL_FEATURE_COLUMNS].isna().sum().rename("Missing rows").to_frame()
````

**Conclusion.** Feature creation keeps one row per matched order. Days since first order stays missing when the first-order date is unknown or later than the labelled order, and value per item stays missing without a valid denominator.

## 4️⃣Split and Verify the Model Dataset

### 4.1 Sort and split by order date

Sort the model-ready orders from oldest to newest. Use approximately 70% for training, 15% for validation, and 15% for testing. The dates at the 70% and 85% row positions start validation and testing. All orders on a boundary day go into the later partition, so percentages can differ slightly.

For retraining, run the same function on the combined, cleaned historical and incoming records. This recalculates the cutoffs for that dataset version.

#### Why use training, validation, and test sets?

- **Training** uses the oldest orders to fit the model and any preprocessing that learns from data, such as imputation or scaling.
- **Validation** uses the next period to compare models, tune settings, and choose the fraud threshold.
- **Test** uses the newest period to estimate performance after those choices are fixed. With only train and test, tuning repeatedly on the test results would make the reported score too optimistic.

#### Split choice and performance estimate

| Approach | What the score estimates | Fit for this project |
|---|---|---|
| Ordinary shuffled K-fold | Average performance on held-out rows drawn from a mixture of dates. | Later orders can appear in training while earlier orders are evaluated. This can hide the effect of changing fraud patterns and make the score optimistic for future orders. |
| Chronological 70/15/15 holdout | Performance on one recent period after the model and threshold are chosen on an earlier validation period. | Matches the order in which the model will see data. Each model configuration can be evaluated with one fit on the training partition, which helps keep automated retraining within the demo time budget. |
| Forward-chaining cross-validation | Performance across several later periods, with each fold trained only on earlier orders. | Shows whether results are consistent across time. It fits each configuration once per fold, so we can use it within the training period if we need a stability check and have time for the extra fits. |

We will use the chronological holdout as the team's shared split. The test score estimates how well the final model would have scored orders in the newest held-out period. A single period can be unusually easy or difficult, so we will report its dates and fraud rate alongside precision and recall. Precision tells us how many flagged orders were actually fraudulent, and recall tells us how many fraudulent orders we caught.

#### Can we spot spikes or trends?

Yes. Keeping the dates lets us compare daily or weekly order volume and fraud rates. Around a sale such as Black Friday, the fraud count could rise simply because more orders were placed. Comparing both order volume and the percentage of orders labelled as fraud with nearby days helps tell those patterns apart. The split summary shows differences across periods. A daily or weekly chart would show when any spikes occurred.

````python
ORDER_KEY_COLUMNS = ["country_code", "order_id"]


def split_orders_by_date(model_data):
    """Recalculate approximate 70/15/15 splits while keeping each day together."""
    # Verify the fields needed for a dated, labelled split.
    required_columns = {*ORDER_KEY_COLUMNS, "order_date", "is_fraud"}
    missing_columns = required_columns.difference(model_data.columns)
    if missing_columns:
        raise ValueError(f"Missing split columns: {sorted(missing_columns)}")
    if not model_data["is_fraud"].isin([0, 1]).all():
        raise ValueError("Fraud labels must be 0 or 1 before splitting.")

    # Check that each order has a date and a unique composite key.
    if model_data["order_date"].isna().any():
        raise ValueError("Every order must have a date before splitting.")
    if model_data[ORDER_KEY_COLUMNS].isna().any().any() or model_data.duplicated(ORDER_KEY_COLUMNS).any():
        raise ValueError("Order keys must be complete and unique before splitting.")

    # Order keys break ties so the assignments are reproducible even if input rows are shuffled.
    sorted_orders = model_data.sort_values(["order_date", *ORDER_KEY_COLUMNS])
    order_days = sorted_orders["order_date"].dt.normalize()
    if order_days.nunique() < 3:
        raise ValueError("At least three distinct order days are needed for three partitions.")

    # Use the dates at the 70% and 85% row positions as the cutoff days.
    validation_start = order_days.iloc[int(len(sorted_orders) * 0.70)]
    test_start = order_days.iloc[int(len(sorted_orders) * 0.85)]
    # Put all orders on each cutoff day into the later partition.
    partitions = {
        "train": sorted_orders.loc[order_days.lt(validation_start)],
        "validation": sorted_orders.loc[order_days.ge(validation_start) & order_days.lt(test_start)],
        "test": sorted_orders.loc[order_days.ge(test_start)],
    }
    if any(partition.empty for partition in partitions.values()):
        raise ValueError("The 70% and 85% cutoffs need distinct days and three nonempty partitions.")

    # Record the partition for each order key so the team can reuse the split.
    assignments = pd.concat([
        partition[[*ORDER_KEY_COLUMNS, "order_date"]].assign(partition=name)
        for name, partition in partitions.items()
    ], ignore_index=True)
    # Summarise the date range, size, and fraud rate of each partition.
    summary = pd.DataFrame.from_dict({
        name: {
            "Start date": partition["order_date"].min().date(),
            "End date": partition["order_date"].max().date(),
            "Rows": len(partition),
            "Share (%)": round(100 * len(partition) / len(sorted_orders), 2),
            "Fraud rows": int(partition["is_fraud"].sum()),
            "Fraud (%)": round(100 * partition["is_fraud"].mean(), 2),
        }
        for name, partition in partitions.items()
    }, orient="index")
    # A one-class partition cannot support the planned fraud evaluation.
    if not summary["Fraud rows"].between(1, summary["Rows"] - 1).all():
        raise ValueError("Each split must contain both fraud and non-fraud orders.")
    return partitions, assignments, summary


def save_split_assignments(assignments, dataset_version, output_folder):
    """Save the shared order assignments and date boundaries for one dataset version."""
    output_folder = Path(output_folder)
    output_folder.mkdir(parents=True, exist_ok=True)
    versioned_assignments = assignments.assign(dataset_version=dataset_version)
    # Get the first and last order dates for each partition in this dataset version.
    boundaries = versioned_assignments.groupby(
        ["dataset_version", "partition"], sort=False
    )["order_date"].agg(first_order_date="min", last_order_date="max")
    versioned_assignments.to_csv(output_folder / f"{dataset_version}_split_assignments.csv", index=False)
    boundaries.to_csv(output_folder / f"{dataset_version}_split_boundaries.csv")


partitions, split_assignments, split_summary = split_orders_by_date(model_data)
# Verify full coverage, unique keys, and chronological order.
assert len(split_assignments) == len(model_data)
assert not split_assignments.duplicated(ORDER_KEY_COLUMNS).any()
assert split_summary.loc["train", "End date"] < split_summary.loc["validation", "Start date"]
assert split_summary.loc["validation", "End date"] < split_summary.loc["test", "Start date"]
split_summary
````

**Conclusion.** The base batch splits into 69.56% training, 15.26% validation, and 15.17% testing. All three partitions contain both label classes and their date ranges are disjoint.

### 4.3 Train, choose, and test

1. **Train on the oldest orders.** Fit the model and learn any preprocessing settings from these orders. For example, if we fill missing `value_per_item` with the median, calculate that median from training orders and reuse it for validation and test.
2. **Choose on validation.** Compare model settings and choose the fraud threshold using the next period. A lower threshold can catch more fraud and send more genuine orders for review.
3. **Check on test.** Once the model settings and threshold are fixed, score the newest held-out orders. This estimates how the chosen model performs on a later period.

**When we retrain**, compare the current model and the newly trained model on the same test orders. Use only orders that neither model used for training or tuning. If there are too few shared test orders, wait for more labelled orders before comparing them.

### 4.5 Fit and verify shared preprocessing

We use training medians to fill missing numeric values. For text features, we learn categories from training and create one indicator column per category. An unseen category has zeros across that field's indicators. Numeric values stay in their original units. Models that need scaling can fit a scaler on training data.

The two-column order key identifies each feature row and its fraud label. Country code is also available as a feature. We check that every partition keeps its original orders and labels, uses the same feature columns in the same order, and has no missing feature values.

````python
def prepare_model_partitions(partitions, feature_columns):
    """Fit on training orders and apply the same rules to each partition."""
    if {"is_fraud", "order_id", "customer_id"}.intersection(feature_columns):
        raise ValueError("Exclude the target and record identifiers from features.")

    training_features = partitions["train"][feature_columns]
    numeric_columns = training_features.select_dtypes(include="number").columns
    training_medians = training_features[numeric_columns].median()
    if training_medians.isna().any():
        raise ValueError("A numeric feature has no training values to learn a median from.")

    category_levels = {
        column: sorted(training_features[column].dropna().unique())
        for column in feature_columns if column not in numeric_columns
    }
    prepared = {}
    for name, orders in partitions.items():
        # Keep the order key for alignment and country code as a model feature.
        indexed = orders.set_index(ORDER_KEY_COLUMNS, drop=False)
        numeric = indexed[numeric_columns].astype("float64").fillna(training_medians)
        categories = pd.DataFrame({
            column: pd.Categorical(indexed[column], categories=levels)
            for column, levels in category_levels.items()
        }, index=indexed.index)
        indicators = pd.get_dummies(categories, dtype="uint8")
        features = pd.concat([numeric, indicators], axis=1)
        target = indexed["is_fraud"].copy()

        expected_keys = pd.MultiIndex.from_frame(orders[ORDER_KEY_COLUMNS])
        if (
            not features.columns.is_unique
            or not features.index.is_unique
            or not features.index.equals(expected_keys)
            or not target.index.equals(expected_keys)
            or not target.reset_index(drop=True).equals(orders["is_fraud"].reset_index(drop=True))
            or features.isna().any().any()
            or "is_fraud" in features.columns
        ):
            raise ValueError(f"Preprocessing validation failed for {name}.")
        prepared[name] = features, target

    training_columns = prepared["train"][0].columns
    if any(not features.columns.equals(training_columns) for features, _ in prepared.values()):
        raise ValueError("Feature names or order differ across partitions.")
    return (training_medians, category_levels), prepared


fitted_preprocessing, prepared_partitions = prepare_model_partitions(partitions, MODEL_FEATURE_COLUMNS)
preprocessing_summary = pd.DataFrame.from_dict({
    name: {
        "Feature rows": len(features),
        "Feature columns": features.shape[1],
        "Target rows": len(target),
        "Missing feature values": int(features.isna().sum().sum()),
    }
    for name, (features, target) in prepared_partitions.items()
}, orient="index")
preprocessing_summary
````

**Conclusion.** All three partitions have the same 36 feature columns and zero missing feature values after preprocessing. Each order keeps its original fraud label in a separate target.
